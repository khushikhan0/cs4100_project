```python
import pandas as pd
```

```python
# Load datasets
df_low = pd.read_csv('low_popularity_spotify_data.csv', low_memory=False)
df_high = pd.read_csv('high_popularity_spotify_data.csv', low_memory=False)
df_apr = pd.read_csv('SpotifyAudioFeaturesApril2019.csv', low_memory=False)
df_nov = pd.read_csv('SpotifyAudioFeaturesNov2018.csv', low_memory=False)

print("Shapes:")
print("Low:", df_low.shape)
print("High:", df_high.shape)
print("April 2019:", df_apr.shape)
print("Nov 2018:", df_nov.shape)
```

```python
for df in [df_low, df_high, df_apr, df_nov]:
    df.columns = df.columns.str.strip().str.lower()
```

```python
for df in [df_low, df_high, df_apr, df_nov]:
    df.drop(columns=df.columns[df.columns.str.contains("^unnamed")],
            inplace=True)
```

```python
for df in [df_low, df_high, df_apr, df_nov]:
    if "track_name" in df.columns:
        df["track_name"] = df["track_name"].astype(str).str.strip().str.lower()
    if "artist_name" in df.columns:
        df["artist_name"] = df["artist_name"].astype(str).str.strip().str.lower()
```

```python
df_low["source"] = "low_popularity"
df_high["source"] = "high_popularity"
df_apr["source"] = "april_2019"
df_nov["source"] = "nov_2018"
```

```python
combined_df = pd.concat(
    [df_low, df_high, df_apr, df_nov],
    ignore_index=True,
    sort=False
)

print("Combined dataset shape:", combined_df.shape)
```

```python
print(combined_df["source"].value_counts())
```

```python
numeric_cols = ["valence", "energy", "tempo", "danceability", "loudness"]

for col in numeric_cols:
    if col in combined_df.columns:
        combined_df[col] = pd.to_numeric(combined_df[col], errors="coerce")
```

```python
if "valence" in combined_df.columns:
    print("Invalid valence:", ((combined_df["valence"] < 0) | (combined_df["valence"] > 1)).sum())
```

```python
print(combined_df.shape)
combined_df = combined_df.drop_duplicates()
print(combined_df.shape)
```

```python
combined_df.to_csv("spotify_master_clean.csv", index=False)
```
